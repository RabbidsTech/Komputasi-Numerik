```python
# ======================================================================
# CELL 2: FUNGSI UTAMA ALGORITMA METODE BISECTION
# ======================================================================
# Identitas: Kresna Tri Samudra (NIM: 202511420050)

def cetak_header_tabel():
    """Mencetak garis pembatas dan judul kolom tabel iterasi Bisection."""
    print("=" * 80)
    print(f"{'Iter':<6} | {'a':<11} | {'b':<11} | {'c (Tengah)':<12} | {'f(c)':<12} | {'Galat':<10}")
    print("=" * 80)

def metode_bisection(f, a, b, toleransi=1e-6, max_iter=100, tampilkan_tabel=True):
    """Mencari akar persamaan f(x) = 0 pada interval [a, b] menggunakan metode Bisection."""
    # Menghitung nilai fungsi pada batas bawah (a) dan batas atas (b)
    fa = f(a)
    fb = f(b)

    # Validasi Teorema Nilai Antara: tanda f(a) dan f(b) wajib berlawanan (hasil kali < 0)
    if fa * fb >= 0:
        print("Error: f(a) dan f(b) harus berlawanan tanda (f(a) * f(b) < 0).")
        return None

    # Menampilkan header tabel jika parameter tampilkan_tabel bernilai True
    if tampilkan_tabel:
        cetak_header_tabel()

    # Melakukan perulangan pencarian akar hingga batas maksimum iterasi
    for iterasi in range(1, max_iter + 1):
        # 1. Menghitung titik tengah (midpoint) dari interval [a, b]
        c = (a + b) / 2.0
        # 2. Mengevaluasi nilai fungsi pada titik tengah c
        fc = f(c)
        # 3. Menghitung galat hampiran berdasarkan setengah panjang interval saat ini
        galat = abs(b - a) / 2.0

        # Mencetak perkembangan nilai pada setiap langkah iterasi
        if tampilkan_tabel:
            print(f"{iterasi:<6} | {a:<11.6f} | {b:<11.6f} | {c:<12.6f} | {fc:<12.6f} | {galat:<10.6f}")

        # 4. Mengecek kriteria berhenti: jika |f(c)| atau galat sudah lebih kecil dari toleransi
        if abs(fc) < toleransi or galat < toleransi:
            if tampilkan_tabel:
                print("=" * 80)
            print(f"\nKonvergen pada iterasi ke-{iterasi}.")
            return c

        # 5. Memperbarui interval [a, b] dengan mempertahankan sub-interval yang mengurung akar
        if fa * fc < 0:
            # Akar berada di antara a dan c, geser batas atas b ke c
            b = c
            fb = fc
        else:
            # Akar berada di antara c dan b, geser batas bawah a ke c
            a = c
            fa = fc

    # Jika loop selesai sebelum toleransi terpenuhi, tampilkan peringatan
    if tampilkan_tabel:
        print("=" * 80)
    print("\nPeringatan: Batas maksimum iterasi tercapai.")
    return c
```

```python
# ======================================================================
# CELL 3: CONTOH PENGGUNAAN, VERIFIKASI SCIPY, DAN VISUALISASI GRAFIK
# ======================================================================

# Mengimpor pustaka NumPy, Matplotlib untuk grafik, dan SciPy untuk verifikasi akar
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import bisect

# Mendefinisikan persamaan non-linear f(x) = x^3 - 4x - 9
def fungsi_contoh(x):
    return x**3 - 4*x - 9

# Menentukan interval awal [2, 3] (karena f(2) = -9 dan f(3) = +6) serta toleransi 10^-6
batas_a = 2.0
batas_b = 3.0
eps = 1e-6

# Menjalankan fungsi metode_bisection buatan sendiri
akar = metode_bisection(fungsi_contoh, batas_a, batas_b, toleransi=eps)

if akar is not None:
    # Menampilkan hasil akar hampiran dan nilai fungsi pada titik akar
    print(f"Akar hampiran (Bisection) : x = {akar:.6f}")
    print(f"Nilai f(x) pada akar      : f({akar:.6f}) = {fungsi_contoh(akar):.8f}")

    # Verifikasi akurasi menggunakan fungsi bawaan scipy.optimize.bisect
    akar_scipy = bisect(fungsi_contoh, batas_a, batas_b, xtol=eps)
    print(f"Verifikasi scipy.bisect   : x = {akar_scipy:.6f}")

    # Membuat data titik x dan y untuk menggambar kurva fungsi f(x)
    x_vals = np.linspace(1.5, 3.5, 400)
    y_vals = fungsi_contoh(x_vals)

    # Menampilkan plot grafik fungsi beserta titik perpotongan akar pada sumbu-x
    plt.figure(figsize=(8, 4.5))
    plt.plot(x_vals, y_vals, label="f(x) = x^3 - 4x - 9", color="navy", linewidth=2)
    plt.axhline(0, color="black", linestyle="--", linewidth=1)
    plt.axvline(akar, color="gray", linestyle=":", linewidth=1)
    plt.scatter([akar], [fungsi_contoh(akar)], color="red", zorder=5, s=60, label=f"Akar x = {akar:.6f}")
    plt.title("Grafik Pencarian Akar Metode Bisection")
    plt.xlabel("Nilai x")
    plt.ylabel("Nilai f(x)")
    plt.legend()
    plt.grid(True, alpha=0.4)
    plt.show()
```

```python
# ======================================================================
# CELL 4: MODE INPUT INTERAKTIF PERSAMAAN POLINOMIAL DERAJAT 3
# ======================================================================
# Bentuk umum persamaan: p3*x^3 + p2*x^2 + p1*x + p0 = 0

print("Masukkan koefisien persamaan p3*x^3 + p2*x^2 + p1*x + p0 = 0")
# Membaca masing-masing koefisien dari input pengguna
p3 = float(input("Koefisien x^3 (p3) : "))
p2 = float(input("Koefisien x^2 (p2) : "))
p1 = float(input("Koefisien x   (p1) : "))
p0 = float(input("Konstanta     (p0) : "))

# Membaca interval awal [a, b] dan batas toleransi error
a_in = float(input("Batas bawah (a)    : "))
b_in = float(input("Batas atas  (b)    : "))
tol_in = float(input("Toleransi (misal 1e-5): "))

# Mendefinisikan fungsi dinamis berdasarkan koefisien yang dimasukkan pengguna
f_user = lambda x: p3*(x**3) + p2*(x**2) + p1*x + p0

# Menjalankan metode Bisection untuk persamaan input pengguna
akar_user = metode_bisection(f_user, a_in, b_in, toleransi=tol_in)

if akar_user is not None:
    print(f"Solusi akar persamaan Anda: x = {akar_user:.6f}")
```
