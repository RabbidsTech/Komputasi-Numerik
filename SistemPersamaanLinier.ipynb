```python
""" CELL 1 - FUNGSI BANTU : MENCETAK MATRIKS AUGMENTED """

def cetak_matriks(M, judul=""):
    """Mencetak matriks augmented [A | b] dengan rapi ke layar."""
    # Mengecek apakah parameter 'judul' diisi; jika ada, cetak sebagai keterangan langkah
    if judul:
        print(judul)
    # Melakukan perulangan untuk membaca setiap baris di dalam matriks M
    for baris in M:
        # baris[:-1] mengambil seluruh elemen koefisien A (kolom awal hingga sebelum terakhir)
        # f"{v:9.4f}" memformat angka agar rata kanan dengan 4 angka di belakang koma
        # baris[-1] mengambil elemen terakhir pada baris sebagai vektor konstanta b
        print("  [" + "  ".join(f"{v:9.4f}" for v in baris[:-1]) + "  |  " + f"{baris[-1]:9.4f}" + " ]")
    # Mencetak baris kosong sebagai pemisah antar langkah matriks
    print()
```

```python
""" CELL 2 - FUNGSI UTAMA ELIMINASI GAUSS-JORDAN """

def gauss_jordan(A, b, tampilkan_langkah=True):
    # Menghitung jumlah persamaan/variabel (n) berdasarkan jumlah baris matriks A
    n = len(A)
    # Membentuk matriks augmented [A | b] berukuran n x (n+1) dan mengubah semua nilai ke tipe float
    M = [list(map(float, A[i])) + [float(b[i])] for i in range(n)]

    # Jika tampilkan_langkah=True, tampilkan bentuk matriks augmented sebelum diproses
    if tampilkan_langkah:
        cetak_matriks(M, "Matriks augmented awal:")

    # Loop utama iterasi per kolom (dari kolom 0 sampai n-1) untuk memproses setiap elemen diagonal (pivot)
    for kol in range(n):
        # 1. Pivoting parsial: mencari baris 'r' (dari baris 'kol' ke bawah) dengan nilai mutlak terbesar di kolom 'kol'
        pivot = max(range(kol, n), key=lambda r: abs(M[r][kol]))
        # Jika nilai pivot terbesar mendekati nol (< 1e-12), matriks bersifat singular dan tidak punya solusi tunggal
        if abs(M[pivot][kol]) < 1e-12:
            print("Matriks singular: tidak ada solusi tunggal.")
            return None

        # Jika baris dengan nilai terbesar bukan baris diagonal saat ini, tukar posisi baris tersebut
        if pivot != kol:
            M[kol], M[pivot] = M[pivot], M[kol]
            if tampilkan_langkah:
                cetak_matriks(M, f"Tukar R{kol+1} <-> R{pivot+1}:")

        # 2. Normalisasi: menyimpan nilai pivot diagonal ke variabel 'p'
        p = M[kol][kol]
        # Membagi seluruh elemen pada baris pivot dengan 'p' sehingga elemen pivot diagonal bernilai 1
        M[kol] = [v / p for v in M[kol]]
        if tampilkan_langkah:
            cetak_matriks(M, f"R{kol+1} = R{kol+1} / {p:.4f}:")

        # 3. Eliminasi semua baris lain (baik di atas maupun di bawah baris pivot)
        for i in range(n):
            # Proses eliminasi hanya diterapkan pada baris selain baris pivot itu sendiri (i != kol)
            if i != kol:
                # Menyimpan nilai pengali (faktor) pada kolom 'kol' yang akan dinolkan
                faktor = M[i][kol]
                # Operasi baris elementer dijalankan jika nilai faktor tidak sudah bernilai nol
                if abs(faktor) > 1e-12:
                    # Mengurangi setiap elemen kolom 'j' pada baris 'i' dengan (faktor * elemen baris pivot)
                    M[i] = [M[i][j] - faktor * M[kol][j] for j in range(n + 1)]
                    if tampilkan_langkah:
                        cetak_matriks(M, f"R{i+1} = R{i+1} - ({faktor:.4f}) * R{kol+1}:")

    # Mengembalikan kolom terakhir matriks (indeks ke-n) sebagai list solusi [x1, x2, ..., xn]
    return [M[i][n] for i in range(n)]
```

```python
""" CELL 3 - CONTOH PENGGUNAAN PADA SPL 3 VARIABEL """

# Mendefinisikan matriks koefisien A (3x3) dari sistem persamaan linear 3 variabel
A = [[ 2,  1, -1],
     [-3, -1,  2],
     [-2,  1,  2]]
# Mendefinisikan vektor konstanta ruas kanan b (3x1)
b = [8, -11, -3]

# Memanggil fungsi gauss_jordan untuk menghitung solusi dan menampilkan langkah OBE
solusi = gauss_jordan(A, b)

# Jika solusi ditemukan (tidak None), tampilkan nilai x1, x2, dan x3 ke layar
if solusi:
    print("Solusi:")
    for i, x in enumerate(solusi, start=1):
        print(f"  x{i} = {x:.4f}")
```

```python
""" CELL 4 - VERIFIKASI HASIL DENGAN NUMPY """

# Mengimpor library NumPy untuk melakukan pengecekan perkalian matriks
import numpy as np

if solusi:
    # Menghitung perkalian matriks A dengan vektor solusi x (A @ x)
    Ax = np.array(A) @ np.array(solusi)
    # Menampilkan hasil perhitungan A @ x yang dibulatkan 6 angka di belakang koma
    print("A @ x      =", np.round(Ax, 6))
    # Menampilkan vektor konstanta b asli sebagai pembanding
    print("b          =", b)
    # Memeriksa apakah hasil A @ x sama persis dengan b menggunakan np.allclose
    print("Cocok?     =", np.allclose(Ax, b))
    # Membandingkan hasil metode Gauss-Jordan buatan sendiri dengan fungsi bawaan np.linalg.solve
    print("numpy.solve =", np.round(np.linalg.solve(A, b), 6))
```

```python
""" CELL 5 - INPUT SENDIRI (INTERAKTIF) """

# Meminta pengguna memasukkan jumlah variabel/persamaan (n)
n = int(input("Jumlah variabel/persamaan: "))
# Menyiapkan list kosong untuk menampung matriks koefisien A_in dan vektor konstanta b_in
A_in, b_in = [], []

# Perulangan untuk meminta input koefisien dan konstanta pada setiap persamaan
for i in range(n):
    # Membaca input baris koefisien yang dipisahkan spasi
    baris = input(f"Koefisien persamaan {i+1} (pisahkan spasi, {n} angka): ").split()
    # Mengubah setiap angka koefisien menjadi float dan menyimpannya ke list A_in
    A_in.append([float(v) for v in baris])
    # Membaca nilai konstanta persamaan ke-(i+1) sebagai float dan menyimpannya ke list b_in
    b_in.append(float(input(f"Konstanta persamaan {i+1}: ")))

# Menjalankan fungsi gauss_jordan tanpa menampilkan langkah-langkah matriks
hasil = gauss_jordan(A_in, b_in, tampilkan_langkah=False)
# Menampilkan hasil solusi akhir jika ditemukan
if hasil:
    print("\nSolusi:")
    for i, x in enumerate(hasil, start=1):
        print(f"  x{i} = {x:.4f}")
```

Metode Gauss-Jordan

Eliminasi Gauss-Jordan menyelesaikan sistem persamaan linear Ax = b dengan mengubah matriks augmented [A | b] melalui operasi baris elementer (OBE) sampai bagian kiri menjadi matriks identitas. Setelah itu, kolom paling kanan langsung berisi solusi tanpa substitusi mundur.

[ a11 a12 a13 | b1 ]         [ 1 0 0 | x1 ]
[ a21 a22 a23 | b2 ]   -->   [ 0 1 0 | x2 ]
[ a31 a32 a33 | b3 ]         [ 0 0 1 | x3 ]

Tiga OBE yang dipakai tidak mengubah himpunan solusi sistem:

Menukar dua baris.
Mengalikan satu baris dengan konstanta tak nol.
Menambah satu baris dengan kelipatan baris lain.

Untuk setiap kolom, langkahnya sebagai berikut:

Pilih elemen pivot di diagonal.
Jadikan pivot bernilai 1 dengan membagi seluruh barisnya dengan nilai pivot.
Nolkan semua elemen lain di kolom itu, baik di atas maupun di bawah pivot.

Perbedaan dengan eliminasi Gauss biasa: Gauss hanya menolkan elemen di bawah diagonal (bentuk segitiga atas) lalu butuh substitusi mundur, sedangkan Gauss-Jordan menolkan elemen di atas dan di bawah sehingga hasilnya langsung terbaca.

Penjelasan Kode

1. Membentuk matriks augmented

python
M = [list(map(float, A[i])) + [float(b[i])] for i in range(n)]

Setiap baris A digabung dengan elemen b yang bersesuaian, dan semuanya diubah ke float agar pembagian akurat.

2. Loop utama per kolom (for kol in range(n))
Setiap iterasi memproses satu kolom, dengan elemen M[kol][kol] sebagai pivot.

3. Pivoting parsial

python
pivot = max(range(kol, n), key=lambda r: abs(M[r][kol]))

Kode mencari baris dengan nilai mutlak terbesar di kolom tersebut, lalu menukarnya ke posisi diagonal. Ini menghindari pembagian dengan nol dan mengurangi galat pembulatan. Jika nilai pivot terbesar pun mendekati nol (< 1e-12), matriks singular dan tidak ada solusi tunggal.

4. Normalisasi

python
M[kol] = [v / p for v in M[kol]]

Seluruh baris pivot dibagi p sehingga elemen pivot menjadi 1.

5. Eliminasi

python
M[i] = [M[i][j] - faktor * M[kol][j] for j in range(n + 1)]

Untuk setiap baris lain (i != kol), kode mengurangi baris itu dengan faktor kali baris pivot, sehingga elemen di kolom tersebut menjadi 0. Kondisi i != kol inilah yang membedakannya dari Gauss biasa, karena baris di atas pivot juga ikut dieliminasi.

6. Hasil
Setelah semua kolom diproses, kolom terakhir (M[i][n]) adalah solusi.

Parameter tampilkan_langkah=True mencetak matriks di setiap langkah, jadi Anda bisa mengikuti prosesnya seperti menghitung manual. Ubah menjadi False jika hanya butuh hasil akhir.

Catatan
Kode ini untuk sistem dengan jumlah persamaan sama dengan jumlah variabel (matriks persegi) dan solusi tunggal.
Kompleksitasnya O(n^3), sehingga cocok untuk ukuran kecil sampai menengah. Untuk sistem besar, numpy.linalg.solve lebih efisien.
Metode ini juga bisa dipakai untuk mencari invers matriks, yaitu dengan menggabungkan A dengan matriks identitas, [A | I], lalu mereduksinya menjadi [I | A^-1].
